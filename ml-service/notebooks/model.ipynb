```python
import os
import shutil
import random
from tqdm import tqdm

SOURCE = DATASET_PATH
DEST = "/content/data"

print("🚀 Starting dataset split...\n")

total_images = 0

# Count total images (for overall progress)
for class_name in os.listdir(SOURCE):
    class_path = os.path.join(SOURCE, class_name)
    total_images += len(os.listdir(class_path))

print(f"📊 Total images: {total_images}\n")

processed_images = 0

for class_name in os.listdir(SOURCE):
    class_path = os.path.join(SOURCE, class_name)
    images = os.listdir(class_path)

    random.shuffle(images)

    total = len(images)

    train = images[:int(0.7 * total)]
    val = images[int(0.7 * total):int(0.85 * total)]
    test = images[int(0.85 * total):]

    print(f"\n📂 Processing: {class_name} ({total} images)")

    for split, data in zip(["train", "val", "test"], [train, val, test]):
        dest_path = os.path.join(DEST, split, class_name)
        os.makedirs(dest_path, exist_ok=True)

        print(f"➡️ {split}: {len(data)} images")

        for img in tqdm(data, desc=f"{class_name}-{split}", leave=False):
            shutil.copy(
                os.path.join(class_path, img),
                os.path.join(dest_path, img)
            )

            processed_images += 1

        print(f"✔ Done {split} for {class_name}")

    print(f"✅ Completed class: {class_name}")

print("\n🎉 ALL DATA SPLIT COMPLETED!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import transforms, models
from torch.utils.data import DataLoader
from torchvision.datasets import ImageFolder
from tqdm import tqdm
from PIL import Image
import os

# =========================
# SAFE DATASET (SKIP BAD IMAGES)
# =========================
class SafeImageFolder(ImageFolder):
    def __getitem__(self, index):
        try:
            return super().__getitem__(index)
        except Exception:
            print(f"⚠️ Skipping bad image: {self.samples[index][0]}")
            return self.__getitem__((index + 1) % len(self.samples))


# =========================
# DEVICE
# =========================
device = torch.device("cuda")
print("🚀 Using GPU:", device)


# =========================
# TRANSFORMS
# =========================
train_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.3, contrast=0.3),
    transforms.ToTensor()
])

val_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])


# =========================
# DATASETS
# =========================
train_data = SafeImageFolder("/content/data/train", transform=train_transform)
val_data = SafeImageFolder("/content/data/val", transform=val_transform)

train_loader = DataLoader(train_data, batch_size=32, shuffle=True)
val_loader = DataLoader(val_data, batch_size=32)

print(f"✅ Dataset loaded ({len(train_data)} training images)")


# =========================
# MODEL (EfficientNet)
# =========================
model = models.efficientnet_b0(weights="DEFAULT")
model.classifier[1] = nn.Linear(model.classifier[1].in_features, len(train_data.classes))
model = model.to(device)

print("✅ Model ready")


# =========================
# LOSS & OPTIMIZER
# =========================
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr=0.0003)


# =========================
# TRAINING LOOP
# =========================
EPOCHS = 5
best_acc = 0

for epoch in range(EPOCHS):
    print(f"\n🔥 Epoch {epoch+1}/{EPOCHS}")

    # ---- TRAIN ----
    model.train()
    train_loss = 0

    for images, labels in tqdm(train_loader, desc="Training"):
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    print(f"✅ Train Loss: {train_loss:.4f}")


    # ---- VALIDATION ----
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in tqdm(val_loader, desc="Validation"):
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)

            _, pred = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (pred == labels).sum().item()

    acc = 100 * correct / total
    print(f"📊 Validation Accuracy: {acc:.2f}%")


    # ---- SAVE BEST MODEL ----
    if acc > best_acc:
        best_acc = acc
        torch.save(model.state_dict(), "best_model.pth")
        print("💾 Best model saved!")


print("\n🎉 Training Complete!")
print(f"🔥 Best Accuracy: {best_acc:.2f}%")
```
